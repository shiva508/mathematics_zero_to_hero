```python
import numpy as np
```

# One-dimensional arrays

```python
heights = np.array([70, 80, 90, 100])
print(heights)
```

# Two-dimensional arrays

```python
two_d_array = np.array([[1, 2], [4, 5]])
print("Shape: ", two_d_array.shape)
print("Type: ", two_d_array.dtype)
print(two_d_array)
```

# Indexing one-dimensional arrays

```python
print("heights[1]: ",heights[1])
print("heights[-1]: ",heights[-1])
```

# Slicing one dimensional arrays

```python
print("heights[0:2]: ",heights[0:2])
print("heights[:2]: ",heights[:2])
print("heights[2:]: ",heights[2:])
print("heights[:]: ",heights[:])
```
