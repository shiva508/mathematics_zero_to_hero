```python
tokens = ['<greeting>', 'Hello World!', '</greeting>']
count = 0

# write your for loop here
for token in tokens:
    if "<" in token or ">" in token or "</" in token:
        count +=1


### Notebook grading
if count == 2:
    print("Nice work!")
else:
    print("Not quite! Did you track of the number of tag with `count`?")
```
