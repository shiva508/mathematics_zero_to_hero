```python
cast_names = ["Barney", "Robin", "Ted", "Lily", "Marshall"]
cast_heights = [72, 68, 72, 66, 76]

def build_cast_ansemble(cast_names,cast_heights):
    cast_ansemble = {}
    for name, height in zip(cast_names, cast_heights):
        cast_ansemble[name]= height
    return cast_ansemble
cast = build_cast_ansemble(cast_names,cast_heights) # replace with your code
print(cast)
```
